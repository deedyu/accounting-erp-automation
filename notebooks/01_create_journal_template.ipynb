# 분개장 양식 및 정상 거래 데이터 생성

가상의 자동차부품 제조기업을 기준으로 ERP 입력 전 단계에서 사용하는
엑셀 분개장 양식과 정상 거래 예시를 생성한다.

## 1. 라이브러리 불러오기 및 저장 경로 설정

```python
# 표 형태의 데이터를 만들고 엑셀로 저장하기 위한 라이브러리
import pandas as pd

# 운영체제와 관계없이 안전하게 파일 경로를 다루기 위한 라이브러리
from pathlib import Path


# 현재 주피터 노트북의 실행 위치 확인
current_path = Path.cwd()

# 노트북이 notebooks 폴더 기준으로 실행되면 상위 폴더를 프로젝트 루트로 설정
if current_path.name == "notebooks":
    project_root = current_path.parent
else:
    project_root = current_path

# 원본 분개장 양식을 저장할 경로 설정
output_path = project_root / "data" / "raw" / "journal_template.xlsx"

print("현재 실행 위치:", current_path)
print("프로젝트 루트:", project_root)
print("파일 저장 위치:", output_path)
```

## 2. 분개장 열 구조 정의

```python
# 인턴 당시 입력 업무를 반영한 원본 분개장 열 정의
journal_columns = [
    "voucher_id",              # 전표번호
    "transaction_date",        # 거래일자
    "department_code",         # 부서 코드
    "partner_code",            # 거래처 코드
    "evidence_type",           # 증빙 유형
    "evidence_no",             # 증빙번호
    "description",             # 적요
    "debit_account_1",         # 첫 번째 차변 계정과목 코드
    "debit_amount_1",          # 첫 번째 차변 금액
    "debit_account_2",         # 두 번째 차변 계정과목 코드
    "debit_amount_2",          # 두 번째 차변 금액
    "credit_account_1",        # 첫 번째 대변 계정과목 코드
    "credit_amount_1",         # 첫 번째 대변 금액
    "credit_account_2",        # 두 번째 대변 계정과목 코드
    "credit_amount_2",         # 두 번째 대변 금액
    "supply_amount",           # 공급가액
    "vat_amount",              # 부가세
    "total_amount",            # 거래 합계금액
    "remarks"                  # 비고
]

# 정의한 열만 포함하는 빈 분개장 데이터프레임 생성
journal_template = pd.DataFrame(columns=journal_columns)

# 생성한 분개장 구조 확인
journal_template
```

## 3. 빈 분개장 양식 저장

```python
# data/raw 폴더가 없을 경우를 대비해 폴더 생성
output_path.parent.mkdir(parents=True, exist_ok=True)

# 빈 분개장 양식을 엑셀 파일로 저장
journal_template.to_excel(
    output_path,
    index=False,
    sheet_name="분개장"
)

# 파일 생성 여부 확인
print("분개장 양식 생성 완료:", output_path.exists())
print("저장된 파일:", output_path)
```

## 5. 차변·대변 및 거래금액 검증

```python
# 전표별 차변 합계 계산
sample_journal["debit_total"] = (
    sample_journal["debit_amount_1"].fillna(0)
    + sample_journal["debit_amount_2"].fillna(0)
)

# 전표별 대변 합계 계산
sample_journal["credit_total"] = (
    sample_journal["credit_amount_1"].fillna(0)
    + sample_journal["credit_amount_2"].fillna(0)
)

# 차변과 대변이 일치하는지 확인
sample_journal["is_balanced"] = (
    sample_journal["debit_total"]
    == sample_journal["credit_total"]
)

# 공급가액과 부가세의 합이 거래금액과 일치하는지 확인
# 계좌이체 같은 정산 거래는 공급가액이 0이므로 검사 대상에서 제외
sample_journal["amount_check"] = (
    (sample_journal["supply_amount"] == 0)
    | (
        sample_journal["supply_amount"]
        + sample_journal["vat_amount"]
        == sample_journal["total_amount"]
    )
)

# 검증 결과 확인
sample_journal[
    [
        "voucher_id",
        "debit_total",
        "credit_total",
        "is_balanced",
        "amount_check"
    ]
]
```

## 6. 정상 분개장 엑셀 파일 저장
